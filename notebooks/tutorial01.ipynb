# Combined CAS and SMILES
EPISuite does not have a database of all CAS numbers. Therefore, we sometimes look up SMILES for a given CAS from other databases such as PubChem and then submit the SMILES to EPISuite for estimating substance properties. Let's create a list of CAS and corresponding SMILES and run it through EPISuite web app. We first submit the CAS RNs to EPISuite and when no result is returned, we submit the SMILES directly:

```python
# define a list of CASRN with corresponding SMILES; the last SMILES is valid with no CAS
cas_smiles_list = [
    {"cas": "50-00-0", "smiles": "C=O"},
    {"cas": "100-00-5", "smiles": "C1=CC=C(C=C1)C=O"},
    {"cas": "100-02-7", "smiles": "C=CC=O"},
    {"cas": "50-17-9", "smiles": "CC(=O)OC1=CC=CC=C1C(=O)O"},  # Aspirin with a valid but wrong CAS
    {"cas": "26402-26-6", "smiles": "CCCCCCCC(=O)OCC(CO)O,CCCCCCCC(=O)OCC(O)CO"},  # valid SMILES with no CAS
]
ids = []
for item in cas_smiles_list:
    res = search_episuite_by_cas([item["cas"]])
    if res:
        ids.extend(res)
    else: # empty list, not found by CAS, search by SMILES
        res = search_episuite([item["smiles"]])
        ids.extend(res)
ids
```

```python
ids = search_episuite_by_cas(["50-17-9"])
if ids:
    print("Found by CAS")
else:
    print("Not found by CAS, searching by SMILES")
```

## Complete Workflow Example
This example demonstrates the full PyEPISuite workflow from search to analysis:

```python
# Complete PyEPISuite workflow example
import pandas as pd
import numpy as np

# Step 1: Define chemicals of interest
chemical_set = {
    "Pharmaceuticals": ["50-78-2", "58-08-2"],  # Aspirin, Caffeine
    "Solvents": ["67-56-1", "64-17-5"],         # Methanol, Ethanol  
    "Industrial": ["100-41-4", "108-88-3"]      # Ethylbenzene, Toluene
}

# Flatten the list
all_chemicals = [cas for category in chemical_set.values() for cas in category]
print(f"Analyzing {len(all_chemicals)} chemicals from {len(chemical_set)} categories")

# Step 2: Search and submit for predictions
ids = search_episuite_by_cas(all_chemicals)
epi_results, ecosar_results = submit_to_episuite(ids)

# Step 3: Convert to DataFrames
epi_df = episuite_to_dataframe(epi_results)
ecosar_df = ecosar_to_dataframe(ecosar_results)

# Add chemical categories
category_map = {}
for category, cas_list in chemical_set.items():
    for cas in cas_list:
        category_map[cas] = category

epi_df['category'] = epi_df['cas'].map(category_map)

print(f"\\nDataFrame created with {len(epi_df)} chemicals and {len(epi_df.columns)} properties")
print(f"Categories: {epi_df['category'].value_counts().to_dict()}")

# Step 4: Basic analysis
key_properties = ['log_kow_estimated', 'water_solubility_logkow_estimated', 
                 'bioconcentration_factor', 'atmospheric_half_life_estimated']

print("\\nKey Properties Summary:")
summary = epi_df.groupby('category')[key_properties].mean().round(2)
print(summary)

# Step 5: Identify chemicals of potential concern
concern_criteria = (
    (epi_df['log_bioconcentration_factor'] > 3) |  # High bioaccumulation
    (epi_df['atmospheric_half_life_estimated'] > 48)  # Persistent
)

chemicals_of_concern = epi_df[concern_criteria]
if len(chemicals_of_concern) > 0:
    print(f"\\nChemicals of potential environmental concern: {len(chemicals_of_concern)}")
    for _, row in chemicals_of_concern.iterrows():
        print(f"  {row['name']} (CAS: {row['cas']}) - Category: {row['category']}")
else:
    print("\\nNo chemicals meet the concern criteria in this dataset")

print("\\n✅ Complete workflow demonstrated successfully!")
```
